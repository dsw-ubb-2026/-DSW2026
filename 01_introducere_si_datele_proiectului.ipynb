### Cum rulăm un notebook
O celulă **Markdown** conține explicații; o celulă **Code** conține Python. Variabilele rămân în memoria kernelului, așa că rulăm celulele de sus în jos. Un `NameError` înseamnă de obicei că ai sărit celula care definea variabila. **Restart** golește memoria; **Run All** rulează totul în ordine.

| Tastă | Ce face |
|---|---|
| **Shift + Enter** | rulează celula și trece la următoarea |
| **Esc** / **Enter** | iese din celulă / intră în celulă pentru a scrie |
| **A** / **B** | adaugă o celulă deasupra / dedesubt |
| **M** / **Y** | transformă celula în Markdown / Code |
| **D, D** | șterge celula |

Tastele **A, B, M, Y, D** funcționează după **Esc**, când celula este selectată, dar nu editată.

## 5. Primul contact cu datele SCI
SCI este construit din prieteniile de pe Facebook și exprimă **probabilitatea relativă ca doi utilizatori din două locuri să fie prieteni**. Valorile sunt scalate: nu sunt procente și nu numără persoane.

Un rând este o **pereche de țări**. `user` și `friend` sunt cele două capete ale perechii, nu o direcție a migrației.

````python
import pandas as pd

# Citim tabelul SCI din folderul data.
# keep_default_na=False: codul "NA" (Namibia) rămâne text, nu devine valoare lipsă.
sci = pd.read_csv("data/country.csv", keep_default_na=False)

sci.head()   # primele 5 rânduri, în ordinea din fișier
````

| Coloană | Ce reprezintă |
|---|---|
| `user_country` | codul primei țări |
| `friend_country` | codul celei de-a doua țări |
| `user_region`, `friend_region` | aici repetă codurile țărilor; în fișierele regionale vor fi, de exemplu, județe |
| `scaled_sci` | indicele de conectivitate socială, scalat |

````python
print(sci.shape)   # (rânduri, coloane)
sci.dtypes         # tipul fiecărei coloane
````
